```python
import numpy as np

# Create a 3 x 2 matrix containing numbers 7 to 12
arr = np.arange(7, 13).reshape(3, 2)

print("Original Matrix:")
print(arr)

print("Shape before transposition:", arr.shape)

# Transpose
transpose = arr.T

print("Transpose:")
print(transpose)

print("Shape after transposition:", transpose.shape)
```
