**Question 21**

**Question 22**

```
s1 = Series({'A': 12, 'B': 18})
s2 = Series({'B': 5, 'C': 9})
print(s1 + s2)
# Addition happens by index label. A is only in s1 and C only in s2,
# so there is no partner value -> NaN. Only B (18+5=23) is in both.
```

**Question 26**

```
north = Series({'Jan': 40, 'Feb': 55, 'Mar': 60})
south = Series({'Feb': 20, 'Mar': 25, 'Apr': 30})
total = north + south
print(total)
# Jan=NaN (only in north), Feb=75, Mar=85, Apr=NaN (only in south)
# Pandas aligns by index label; labels not in both give NaN.
```

**Question 27**

```
d = {'Red': 12, 'Blue': 18, 'Green': 24}
labels = ['Green','Red','Yellow','Blue']
result = Series(d, index=labels)
print(result)
print(pd.isnull(result))
# (a) Yellow has NaN
# (b) pd.isnull(result)['Yellow'] -> True
# (c) 'Yellow' is in the given index but not in the dictionary keys
# (d) Green=24, Red=12, Blue=18 keep numeric values
```

**Question 28**

```
s = Series([90, 75, 82], index=['Ravi','Aman','Neha'])
d = s.to_dict()
s2 = Series(d)
print(d)
print(s2)
# s2 index: Ravi, Aman, Neha; values: 90, 75, 82
# Yes, s and s2 represent the same labelled data.
```

**Question 29**

```
products = Series([450, 620, 380, 710], index=['Pen', 'Bag', 'Book', 'Shoes'])
print(products.values)          # values
print(products.index)           # index
print(products['Bag'])          # select one label
print(products[products > 500]) # filter sales above 500
print('Book' in products.index) # check label exists
print(products.to_dict())       # convert to dictionary
```

**Question 30**

```
data = {'A': 15, 'B': 35, 'C': 25}
s = Series(data, index=['C','A','D','B'])
s.name = 'Observations'
s.index.name = 'Label'   # the question's last statement is cut off; 'Label' assumed
print(s)
print(s[s > 20])        # (c) C=25, B=35
print(pd.notnull(s))    # (d) C True, A True, D False, B True
# (a) Index order: C, A, D, B
# (b) D has NaN
# (e) Series name = 'Observations', index name = 'Label'
```
