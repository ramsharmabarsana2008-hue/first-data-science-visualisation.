## Topic 3: Identity (is) vs Equality (==)

**Q3.1**

```
# == checks whether two objects have the same VALUE.
# 'is' checks whether two names refer to the exact same OBJECT in memory.
x = 3
y = 3.0
print(x == y)
print(x is y)
```

**Q3.2**

```
# Prediction: True, False
a = [1, 2, 3]
b = [1, 2, 3]
print(a == b)
print(a is b)
```

**Q3.3**

```
x = 100
y = 100
print(x is y, x == y)
p = 1000
q = 1000
print(p is q, p == q)
# CPython caches small integers (-5 to 256), so they share one object.
# Larger ints may be separate objects, so 'is' can be False.
# (In a notebook cell, constants may be merged, so p is q can show True too.)
# Always use == to compare values.
```

## Topic 7: Functions & Passing Arguments

**Q7.1**

```
def my_abs(val):
    if val < 0:
        return 0 - val
    return val

print(my_abs(5))
print(my_abs(-7))
print(my_abs(0))
```

**Q7.2**

```
# my_abs("Hi") -> TypeError: '<' not supported between 'str' and 'int'
try:
    my_abs("Hi")
except TypeError as err:
    print("TypeError:", err)
```

**Q7.3**

```
def print_abs(val):
    if val < 0:
        print(0 - val)
    else:
        print(val)

x = print_abs(-2.7)
print(x)
# x is None because the function prints but has no return statement.
```

**Q7.4**

```
# Prediction: 7
def inc_val(val):
    val = val + 1   # only changes the local name 'val'

x = 7
inc_val(x)
print(x)
# Reassigning val makes the local name point to a new int (8).
# x still points to the original 7, so it is unchanged.
```

**Q7.5**

```
def swap_message(val1, val2):
    val1, val2 = val2, val1
    print(val1, val2)

x = 6
y = 3
swap_message(6, 3)   # prints 3 6
print(x, y)          # still 6 3
```

## Topic 8: Scope

**Q8.1**

```
# Prediction: NameError - 'val' exists only inside my_abs()
def my_abs(val):
    if val < 0:
        return 0 - val
    return val

try:
    print(val)
except NameError as err:
    print("NameError:", err)
```

**Q8.2**

```
counter = 0
def show_counter():
    print(counter)

show_counter()
```

**Q8.3**

```
def increment_counter():
    counter = counter + 1

try:
    increment_counter()
except UnboundLocalError as err:
    print("UnboundLocalError:", err)
# Assigning to counter makes it a LOCAL variable, which is read before
# it has any value, hence the error.
```

**Q8.4**

```
def increment_counter():
    global counter
    counter = counter + 1

increment_counter()
print(counter)
```
