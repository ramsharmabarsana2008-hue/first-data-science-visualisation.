## 1. Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

pd.set_option('display.max_columns', None)
%matplotlib inline
```

## 2. Load the Dataset

```python
# Put your downloaded Kaggle CSV in the same folder as this notebook
# (on Google Colab, upload it first using the Files panel on the left)
DATA_PATH = 'dataset.csv'   # <-- change this to your file name

try:
    df = pd.read_csv(DATA_PATH)
except UnicodeDecodeError:
    df = pd.read_csv(DATA_PATH, encoding='latin-1')

df.head()
```

## 3. Dataframe Dimensions (`shape`)

```python
rows, cols = df.shape
print(f"Number of rows    : {rows}")
print(f"Number of columns : {cols}")
print("Shape:", df.shape)
```

## 4. Column Headers (`columns`)

```python
print(df.columns.tolist())
```

## 5. Data Types and Non-Null Counts (`info()`)

```python
df.info()
```

## 6. Summary Statistics (`describe()`)

```python
# Numerical columns
df.describe()
```

```python
# Categorical / text columns
df.describe(include='object')
```

## 7. Additional Checks

```python
# Missing values
missing = df.isnull().sum()
missing_pct = (missing / len(df) * 100).round(2)
pd.DataFrame({'Missing Count': missing, 'Missing %': missing_pct}).sort_values('Missing Count', ascending=False)
```

```python
# Duplicate rows
print("Duplicate rows:", df.duplicated().sum())
```

```python
# Unique values per column
df.nunique().sort_values(ascending=False)
```

## 8. Quick Visualizations

```python
num_cols = df.select_dtypes(include=np.number).columns.tolist()

if num_cols:
    df[num_cols].hist(figsize=(12, 8), bins=20, edgecolor='black')
    plt.suptitle('Distributions of Numerical Columns')
    plt.tight_layout()
    plt.show()
else:
    print("No numerical columns found.")
```

```python
if len(num_cols) > 1:
    corr = df[num_cols].corr()
    plt.figure(figsize=(8, 6))
    plt.imshow(corr, cmap='coolwarm', vmin=-1, vmax=1)
    plt.colorbar(label='Correlation')
    plt.xticks(range(len(num_cols)), num_cols, rotation=90)
    plt.yticks(range(len(num_cols)), num_cols)
    plt.title('Correlation Heatmap')
    plt.tight_layout()
    plt.show()
```
