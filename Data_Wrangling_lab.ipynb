```python
import pandas as pd
import numpy as np
from IPython.display import display

pd.set_option('display.max_columns', None)
pd.set_option('display.width', 120)
```

```python
import zipfile
import io
from google.colab import files
import pandas as pd

uploaded = files.upload()

for filename in list(uploaded.keys()):
    if filename.endswith('.zip'):
        print(f"Extracting {filename}...")
        with zipfile.ZipFile(io.BytesIO(uploaded[filename]), 'r') as zip_ref:
            zip_ref.extractall('.')
            for name in zip_ref.namelist():
                uploaded[name] = True

required_files = [
    'students_master.csv', 'exam_marks.csv', 'semester_enrollments.csv',
    'section_CA.csv', 'section_CB.csv', 'marks_original.csv',
    'marks_updated.csv', 'marks_long.csv', 'repeat_tests.csv'
]

data = {}
for filename in required_files:
    try:
        data[filename] = pd.read_csv(filename)
    except FileNotFoundError:
        raise FileNotFoundError(f"Please upload {filename} or a ZIP containing it.")

students = data['students_master.csv']
exam = data['exam_marks.csv']
enroll = data['semester_enrollments.csv']
ca = data['section_CA.csv']
cb = data['section_CB.csv']
original = data['marks_original.csv']
updated = data['marks_updated.csv']
long_marks = data['marks_long.csv']
repeat = data['repeat_tests.csv']

print("All datasets loaded successfully.")
```

```python
display(students.head())
display(exam.head())
common_columns = students.columns.intersection(exam.columns).tolist()
print("Common columns:", common_columns)
```

```python
q2 = pd.merge(students, exam, on='Student_ID', how='inner')
display(q2)
```

```python
q3 = pd.merge(students, exam, on='Student_ID', how='left')
display(q3)
```

```python
q4 = pd.merge(students, exam, on='Student_ID', how='right')
display(q4)
```

```python
q5 = pd.merge(students, exam, on='Student_ID', how='outer')
display(q5)
print("All Student_ID values:", q5['Student_ID'].tolist())
```

```python
q6 = pd.concat([ca, cb], axis=0)
display(q6)
print("Index:", q6.index.tolist())
```

```python
q7 = pd.concat([ca, cb], axis=0, ignore_index=True)
display(q7)
print("New index:", q7.index.tolist())
```

```python
q8 = pd.concat([ca, cb], axis=1)
display(q8)
print("Column labels:", q8.columns.tolist())
```

```python
attendance_series = pd.Series(ca['Attendance'], name='Attendance')
student_id_series = pd.Series(cb['Student_ID'], name='Student_ID')
q9 = pd.concat([attendance_series, student_id_series], axis=0)
display(q9)
```
