## Question 21: Daily and product sales totals

```python
import numpy as np

sales = np.array([
    [120, 150, 180],
    [140, 160, 175],
    [135, 155, 190],
    [150, 170, 200]
])

daily_totals = sales.sum(axis=1)
product_totals = sales.sum(axis=0)
highest_day = daily_totals.argmax() + 1

print("Total sales for each day:", daily_totals)
print("Total sales for each product:", product_totals)
print("Highest combined sales: Day", highest_day, "with", daily_totals.max())
```

**Explanation:** Rows are days and columns are products. `axis=1` adds across each row (daily totals); `axis=0` adds down each column (product totals). `argmax()` returns a zero-based position, so add 1 for Day 1–4.

## Question 22: Extract a submatrix and calculate its mean

```python
import numpy as np

arr = np.arange(10, 46).reshape(6, 6)
submatrix = arr[2:6, 1:5]

print("Original array:\n", arr)
print("Extracted submatrix:\n", submatrix)
print("Mean:", submatrix.mean())
```

**Explanation:** `np.arange(10, 46)` creates integers 10 through 45. Python slice endpoints are exclusive, so `2:6` selects row indices 2–5 and `1:5` selects column indices 1–4. The extracted matrix is 4 × 4.

## Question 23: Boolean indexing and replacing low weights

```python
import numpy as np

weights = np.array([48, 62, 75, 81, 54, 69, 90])

above_70 = weights[weights > 70]
adjusted = np.where(weights < 55, 55, weights)

print("Weights above 70:", above_70)
print("New array with weights below 55 replaced:", adjusted)
print("Original array unchanged:", weights)
```

**Explanation:** Boolean indexing selects only values greater than 70. `np.where(condition, 55, weights)` creates a new array in which values below 55 become 55; the original `weights` array is unchanged.

## Question 24: Product costs, overall bill, and percentages

```python
import numpy as np

quantity = np.array([3, 5, 2, 7])
price = np.array([120, 80, 250, 60])

costs = quantity * price
bill = costs.sum()
percentages = costs / bill * 100

print("Cost per product:", costs)
print("Overall bill:", bill)
print("Percentage contribution (%):", np.round(percentages, 2))
```

**Explanation:** Multiply each quantity by its matching unit price. Add the costs for the bill total. Each product's contribution is `product cost / overall bill × 100`.

## Question 25: Row-wise and column-wise variance and standard deviation

```python
import numpy as np

arr = np.array([
    [10, 20, 30, 40],
    [12, 22, 32, 42],
    [15, 25, 35, 45]
])

print("Array:\n", arr)
print("Variance by row (axis=1):", np.var(arr, axis=1))
print("Standard deviation by row (axis=1):", np.std(arr, axis=1))
print("Variance by column (axis=0):", np.var(arr, axis=0))
print("Standard deviation by column (axis=0):", np.std(arr, axis=0))
```

**Explanation:** `axis=1` operates across columns within each row, giving one statistic per row. `axis=0` operates down rows within each column, giving one statistic per column. NumPy's default variance and standard deviation use population formulas (`ddof=0`).

## Question 26: Matrix multiplication and element-wise multiplication

```python
import numpy as np

A = np.array([[1, 2, 3],
              [4, 5, 6]])       # shape (2, 3)
B = np.array([[7, 8],
              [9, 10],
              [11, 12]])       # shape (3, 2)

product = A @ B
print("A shape:", A.shape)
print("B shape:", B.shape)
print("Matrix product A @ B:\n", product)
print("Product shape:", product.shape)

# A * B would raise a broadcasting error because (2, 3) and (3, 2)
# are not compatible for element-wise multiplication.
```

**Explanation:** Matrix multiplication is valid because A's number of columns (3) equals B's number of rows (3); the result has shape (2, 2). Element-wise multiplication pairs corresponding positions and requires equal shapes or broadcast-compatible shapes, which `(2, 3)` and `(3, 2)` are not.

## Question 27: Random integers, sorting, five largest, and unique count

## Question 28: View versus independent copy

```python
import numpy as np

a = np.array([5, 10, 15, 20, 25, 30])
b = a[1:5]          # view into a
c = a[1:5].copy()   # independent copy

b[1] = 99
c[2] = 77

print("a:", a)
print("b:", b)
print("c:", c)
```

**Explanation:** Final contents: `a = [5, 10, 99, 20, 25, 30]`, `b = [10, 99, 20, 25]`, and `c = [10, 20, 77, 25]`. A slice like `a[1:5]` is normally a view, so changing `b` changes `a`. `.copy()` makes independent data, so changing `c` does not change `a`.

## Question 29: Add revenue Series and explain alignment

## Question 32: Subtract Series by labels and find non-missing entries

```python
import pandas as pd

s1 = pd.Series([100, 200, 300], index=['North', 'South', 'East'])
s2 = pd.Series([10, 20, 30], index=['South', 'West', 'North'])

result = s1 - s2
print("Result:\n", result)
print("Non-missing entries:\n", result[result.notnull()])
```

**Explanation:** Subtraction aligns matching labels: North = 100 − 30 = 70; South = 200 − 10 = 190. East and West have no matching value in the other Series, so they become NaN. `notnull()` selects non-missing entries.

## Question 35: Examination performance analysis with reproducible random marks

```python
import numpy as np

rng = np.random.default_rng(2026)
marks = rng.integers(20, 101, size=(12, 4))

student_averages = marks.mean(axis=1)
test_means = marks.mean(axis=0)
test_stds = marks.std(axis=0)
passed_all = np.all(marks >= 40, axis=1)
students_average_75 = np.where(student_averages >= 75)[0] + 1
greatest_variation_test = test_stds.argmax() + 1

print("Marks (rows = students 1–12, columns = tests 1–4):\n", marks)
print("Each student's average:", np.round(student_averages, 2))
print("Mean of each test:", np.round(test_means, 2))
print("Standard deviation of each test:", np.round(test_stds, 2))
print("Number passing all four tests:", passed_all.sum())
print("Students averaging at least 75:", students_average_75)
print("Test with greatest variation: Test", greatest_variation_test)
```

**Explanation:** A fixed seed makes the generated marks reproducible. `axis=1` calculates one average per student; `axis=0` calculates one mean or standard deviation per test. Passing all four means every mark is at least 40. Student numbering in the output is 1-based (row index + 1).

## Question 36: Central matrix, transpose, product, and row sums

```python
import numpy as np

matrix = np.arange(21, 46).reshape(5, 5)
central = matrix[1:4, 1:4]
transposed = central.T
product = central @ transposed
row_sums = product.sum(axis=1)
maximum = row_sums.max()
positions = np.argwhere(row_sums == maximum)

print("5 x 5 matrix:\n", matrix)
print("Central 3 x 3 matrix:\n", central)
print("Transpose (3 x 3):\n", transposed)
print("Product central @ transpose (3 x 3):\n", product)
print("Row-wise sums:", row_sums)
print("Maximum row sum:", maximum)
print("Zero-based positions tied at maximum:", positions.tolist())
```

**Explanation:** The original matrix is 5×5. Slicing rows and columns 1–3 extracts a 3×3 center. Its transpose is also 3×3, and multiplying the two gives a 3×3 product. Row sums have length 3. `argwhere` reports zero-based positions; add 1 to convert to 1-based row numbering.

## Question 37: Sales quantities and revenue analysis

```python
import numpy as np

quantities = np.array([
    [12, 8, 20, 5],
    [18, 10, 15, 7],
    [20, 12, 18, 9],
    [16, 14, 22, 8],
    [25, 11, 17, 10],
    [19, 15, 20, 12]
])
original_prices = np.array([150, 240, 90, 320])

daily_revenue = quantities @ original_prices
product_quantity_totals = quantities.sum(axis=0)
product_revenue_totals = quantities.sum(axis=0) * original_prices
highest_day = daily_revenue.argmax() + 1
best_product = product_quantity_totals.argmax() + 1

discounted_prices = np.where(product_quantity_totals > 100,
                             original_prices * 0.90,
                             original_prices)
new_total_revenue = (quantities @ discounted_prices).sum()

print("Daily revenue:", daily_revenue)
print("Total revenue per product:", product_revenue_totals)
print("Overall revenue at original prices:", daily_revenue.sum())
print("Highest-revenue day: Day", highest_day)
print("Total quantity per product:", product_quantity_totals)
print("Best-selling product: Product", best_product)
print("Original prices:", original_prices)
print("Prices after conditional 10% reduction:", discounted_prices)
print("Revenue after price reductions:", new_total_revenue)
```

**Explanation:** This uses a sample 6×4 quantity matrix. Daily revenue is matrix multiplication of quantities by unit prices. Products with total quantity strictly greater than 100 receive a 10% discount. `np.where()` creates discounted prices without changing `original_prices`.

## Question 38: Clean normal observations by clipping to boundaries

```python
import numpy as np

rng = np.random.default_rng(2026)
observations = rng.normal(loc=50, scale=12, size=100)
outside_count = np.sum((observations < 30) | (observations > 70))
cleaned = np.where(observations < 30, 30,
                   np.where(observations > 70, 70, observations))

def summary(x):
    return {
        "mean": x.mean(),
        "variance": x.var(),
        "standard deviation": x.std(),
        "minimum": x.min(),
        "maximum": x.max()
    }

print("Values outside 30–70:", outside_count)
print("Original statistics:", summary(observations))
print("Cleaned statistics:", summary(cleaned))
```

**Explanation:** The seed makes the sample reproducible. Values below 30 are replaced with 30 and values above 70 with 70. This clipping reduces the range and usually reduces variance and standard deviation; the mean can move up or down depending on the sample. The minimum and maximum of the cleaned data will be within 30–70.

## Question 39: Financial reconciliation across branches

```python
import pandas as pd

branch_a = pd.Series({'Q1': 120, 'Q2': 145, 'Q4': 180})
branch_b = pd.Series({'Q2': 110, 'Q3': 135, 'Q4': 155})
branch_c = pd.Series({'Q1': 95, 'Q3': 125, 'Q4': 140})

ordinary = branch_a + branch_b + branch_c
common_quarters = sorted(set(branch_a.index) & set(branch_b.index) & set(branch_c.index))
common_totals = ordinary.reindex(common_quarters)
zero_filled = branch_a.add(branch_b, fill_value=0).add(branch_c, fill_value=0)

print("Ordinary addition (missing if any branch lacks quarter):\n", ordinary)
print("Quarters common to all branches:", common_quarters)
print("Combined earnings for common quarters:\n", common_totals)
print("Absent records treated as zero:\n", zero_filled)
```

**Explanation:** Ordinary addition aligns by quarter label and yields NaN if any branch has no record for that quarter. Only Q4 appears in all three branches, totaling 180 + 155 + 140 = 475. Using `fill_value=0` treats absent records as zero, which assumes no earnings rather than unknown/unrecorded earnings.

## Question 40: Preserve index order and validate student scores

```python
import pandas as pd

scores = {'R101': 78, 'R102': 65, 'R104': 92, 'R105': 48}
required = ['R105', 'R103', 'R101', 'R106', 'R104', 'R102']

s = pd.Series(scores).reindex(required)
missing_records = s[s.isna()].index.tolist()
valid_70_plus = s[s.notna() & (s >= 70)]
filtered_dict = valid_70_plus.to_dict()
reconstructed = pd.Series(filtered_dict)

print("Scores in required order:\n", s)
print("Missing student records:", missing_records)
print("Valid scores at least 70:\n", valid_70_plus)
print("Dictionary:", filtered_dict)
print("Reconstructed Series:\n", reconstructed)
print("Labelled values match:", reconstructed.equals(valid_70_plus))
```

**Explanation:** `reindex(required)` preserves the exact required order and inserts NaN for missing IDs. The missing records are R103 and R106. Filtering first checks scores are not missing and then selects scores ≥70. Missing records should not automatically be treated as zero because no recorded score is different from a score of zero.
