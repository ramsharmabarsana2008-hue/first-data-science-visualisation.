## Question 1: Create `student_df` from a dictionary

## Question 25: Read clipboard data, display columns, and select three

## Question 26: Course DataFrame operations in sequence

```python
course_df = pd.DataFrame({
    'Student': ['Aman', 'Priya', 'Ravi', 'Neha', 'Kabir'],
    'Course': ['Python', 'SQL', 'Java', 'Python', 'Java'],
    'Score': [88, 91, 76, 85, 80]
})

print("Columns:", course_df.columns.tolist())
print("Student column:\n", course_df['Student'])
student_score_df = course_df[['Student', 'Score']]
course_df['Batch'] = '2026'
print("Student and Score:\n", student_score_df)
print("Updated course_df:\n", course_df)
```

**Explanation:** The operations are performed in the requested order: inspect columns, select Student, select two columns, then add Batch.

## Question 27: Travel DataFrame and partially indexed Series

```python
travel_df = pd.DataFrame({
    'Traveller': ['Aman', 'Priya', 'Ravi', 'Neha', 'Kabir'],
    'Destination': ['Delhi', 'Agra', 'Jaipur', 'Pune', 'Mumbai']
})

travel_df['Code'] = np.arange(5)
print("After assigning np.arange(5):")
display(travel_df)

travel_codes = pd.Series(['T1', 'T4'], index=[1, 4])
travel_df['Code'] = travel_codes
print("After assigning partially indexed Series:")
display(travel_df)
```

**Explanation:** The first assignment fills all rows with 0–4. The second aligns values at indexes 1 and 4; the other Code cells become `NaN`.

## Question 28: Create school_df, request House, then fill it

## Question 29: Add a partially indexed Series, then delete it

```python
inventory_df = pd.DataFrame({
    'Item': ['Pen', 'Notebook', 'Eraser', 'Ruler'],
    'Quantity': [20, 15, 30, 12]
})

stock_status = pd.Series(['Reorder', 'Available'], index=[1, 3])
inventory_df['Status'] = stock_status
print("After adding partially indexed Status:")
display(inventory_df)

inventory_df = inventory_df.drop(columns=['Status'])
print("After deleting Status:")
display(inventory_df)
```

**Explanation:** Only index labels 1 and 3 receive Status values. The other rows are `NaN`; `drop(columns=['Status'])` removes the new column.

## Question 30: Clipboard workflow: display, select columns, add, and delete

```python
# Copy a table with at least four columns, including a header row, before running.
try:
    lab_df = pd.read_clipboard()
except Exception as e:
    print("Clipboard reading is unavailable:", e)
    lab_df = pd.DataFrame({
        'Student': ['Aman', 'Priya'],
        'Total Marks': [85, 92],
        'City': ['Delhi', 'Pune'],
        'Department': ['IT', 'HR']
    })

print("Complete DataFrame:")
display(lab_df)
print("Column names:", lab_df.columns.tolist())

# Retrieve the first single-word column, if one exists.
single_word_cols = [col for col in lab_df.columns if isinstance(col, str) and ' ' not in col]
if single_word_cols:
    print("Single-word column", repr(single_word_cols[0]), ":")
    print(lab_df[single_word_cols[0]])
else:
    print("No single-word column found.")

# Retrieve a column containing a space, if one exists.
spaced_cols = [col for col in lab_df.columns if isinstance(col, str) and ' ' in col]
if spaced_cols:
    print("Column containing a space", repr(spaced_cols[0]), ":")
    print(lab_df[spaced_cols[0]])
else:
    print("No column containing a space found.")

# Make a smaller DataFrame from up to three existing columns.
selected_cols = lab_df.columns[:3].tolist()
small_lab_df = lab_df[selected_cols]
print("Smaller DataFrame:")
display(small_lab_df)

lab_df['LabStatus'] = 'Checked'
print("After adding LabStatus:")
display(lab_df)

lab_df = lab_df.drop(columns=['LabStatus'])
print("After deleting LabStatus:")
display(lab_df)
```

**Explanation:** `pd.read_clipboard()` reads the copied table. The code safely finds a column with no spaces and one with a space, selects up to three columns, adds a constant column, and then deletes it. In Google Colab, clipboard access may not work in every browser/session; use the fallback data if needed.
